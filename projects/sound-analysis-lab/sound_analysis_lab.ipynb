```python
#@title 2. 音源と調べる範囲を選ぶ
#@markdown **合成デモ**：toneは正弦波、harmonicsは倍音、bandは複数の楽器を重ねた音です。PCのファイルやYouTubeは、自分に分析・保存する権利がある音源を使います。
source_mode = "合成デモ" #@param ["合成デモ", "PCの音声ファイル", "YouTube URL"]
demo_name = "band" #@param ["tone", "harmonics", "band"]
youtube_url = "" #@param {type:"string"}
#@markdown **全体**を選ぶと開始・終了の入力をリセットしたのと同じ範囲になります。区間指定では**終了0は曲末**です。1回に20分まで、ファイルは250 MBまで扱えます。
range_mode = "全体" #@param ["全体", "区間を指定"]
start_seconds = 0 #@param {type:"number"}
end_seconds = 0 #@param {type:"number"}

# 入力を変更したら、以前の音や分析結果を使わずにやり直す。
lab = StudySession()
input_path = None
if source_mode == "PCの音声ファイル":
    try:
        from google.colab import files
    except ImportError as exc:
        raise RuntimeError("ファイル選択はColabで利用できます。ColabでこのNotebookを開いてください。") from exc
    uploaded = files.upload()
    if len(uploaded) != 1:
        raise ValueError("音声ファイルを1つ選んでください。")
    filename, content = next(iter(uploaded.items()))
    suffix = Path(filename).suffix.lower()
    if len(suffix) > 8 or not suffix[1:].isalnum():
        suffix = ".audio"
    input_path = lab.directory / ("uploaded" + suffix)
    input_path.write_bytes(content)
    del uploaded, content
original = lab.prepare(source_mode=source_mode, demo_name=demo_name, path=input_path,
                       youtube_url=youtube_url, range_mode=range_mode,
                       start=start_seconds, end=end_seconds)
print(f"音源を用意しました：{len(original.data) / original.sr:.2f}秒、{original.sr} Hz、{original.data.shape[1]}チャンネル")
```

```python
#@title 3. パートを用意する
#@markdown **自動**は合成デモの既知のパートを使います。実際の音源はDemucsでボーカル・ドラム・ベース・その他へ分けます。実音源の分離は初回にモデルを取得します。ColabのGPUが使えると処理が速くなります。
separation_mode = "自動" #@param ["自動", "分離しない", "Demucsで4パート分離"]
device = "自動" #@param ["自動", "CPU", "GPU"]

stems = lab.separate(mode=separation_mode, device=device)
print("パート：" + ("、".join(audio.label for audio in stems.values()) or "元の音だけで分析します"))
print("AIによる分離では、別のパートの音や人工的な成分が残る場合があります。")
```

```python
#@title 4. 聴く・見る・比べる
#@markdown 波形は時間と振幅、スペクトルは周波数成分、スペクトログラムはその時間変化を表します。**RMS**は振幅の大きさ、**スペクトル重心**は明るさの目安です。重心は音高そのものではありません。
#@markdown FFTの幅を大きくすると周波数を細かく見られる一方で、短い変化が見えづらくなります。無音しきい値は元の音のピークを共通の基準にし、小さいパートを個別に持ち上げません。
n_fft = 2048 #@param [512, 1024, 2048, 4096, 8192] {type:"raw"}
hop = 512 #@param [128, 256, 512, 1024] {type:"raw"}
silence_db = -50 #@param {type:"slider", min:-80, max:-10, step:1}

result = lab.analyse(n_fft=n_fft, hop=hop, silence_db=silence_db)
show_result(result, accompaniment=True)
```

```python
#@title 5. 加工して変化を確かめる（任意）
#@markdown **高い音を抑える**はローパスフィルターです。**小さい音を弱める**は振幅に応じたゲートです。ゲートはノイズだけを選ぶ処理ではないため、弱い音や余韻も変わります。改善したかどうかは耳とグラフで判断しましょう。
processing_mode = "加工しない" #@param ["加工しない", "高い音を抑える", "小さい音を弱める"]
target_part = "元の音" #@param ["元の音", "ボーカル", "ドラム", "ベース", "その他"]
#@markdown ローパスでは周波数（Hz）、ゲートでは小さい成分を残す割合（0〜1）を使います。0で強く抑え、1ですべて残します。
cutoff_hz = 3000 #@param {type:"number"}
gate_retention = 0.2 #@param {type:"slider", min:0, max:1, step:0.05}

modes = {"加工しない": "none", "高い音を抑える": "lowpass", "小さい音を弱める": "gate"}
parts = {"元の音": "original", **{label: key for key, label in STEM_LABELS.items()}}
processed = lab.process(target=parts[target_part], mode=modes[processing_mode],
                        cutoff=cutoff_hz, strength=gate_retention)
if processed is None:
    print("加工は行いません。セル4の結果をセル6から保存できます。")
else:
    show_result(processed)
```
